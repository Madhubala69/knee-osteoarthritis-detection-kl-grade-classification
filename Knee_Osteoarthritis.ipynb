```python
train_dir = "/kaggle/input/datasets/tommyngx/kneeoa/train"
valid_dir = "/kaggle/input/datasets/tommyngx/kneeoa/val"
test_dir  = "/kaggle/input/datasets/tommyngx/kneeoa/test"

IMG_SIZE = (300,300)
BATCH = 16
EPOCHS_STAGE1 = 50
EPOCHS_STAGE2 = 25
```

```python
def focal_loss(gamma=2., alpha=0.25):
    def loss(y_true, y_pred):
        epsilon = 1e-7
        y_pred = tf.clip_by_value(y_pred, epsilon, 1. - epsilon)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = alpha * tf.pow(1 - y_pred, gamma)
        return tf.reduce_sum(weight * cross_entropy, axis=1)
    return loss
```

```python
def train_model(model_name, base_model_class, preprocess_func):

    import tensorflow.keras.backend as K
    K.clear_session()

    print(f"\n================ {model_name} =================")

    size = (224,224) if model_name == "ResNet50" else IMG_SIZE

    # ---------------- DATA GENERATORS ----------------
    train_gen = ImageDataGenerator(
        preprocessing_function=preprocess_func,
        rotation_range=25,
        zoom_range=0.25,
        horizontal_flip=True,
        width_shift_range=0.1,
        height_shift_range=0.1,
        shear_range=0.1,
        fill_mode="nearest",
        brightness_range=[0.7,1.3]
    ).flow_from_directory(
        train_dir,
        target_size=size,
        batch_size=BATCH,
        class_mode="categorical"
    )

    from sklearn.utils.class_weight import compute_class_weight
    import numpy as np

    class_weights = compute_class_weight(
        class_weight="balanced",
        classes=np.unique(train_gen.classes),
        y=train_gen.classes
    )

    class_weights = dict(enumerate(class_weights))
    print("Class Weights:", class_weights)

    val_gen = ImageDataGenerator(
        preprocessing_function=preprocess_func
    ).flow_from_directory(
        valid_dir,
        target_size=size,
        batch_size=BATCH,
        class_mode="categorical"
    )

    test_gen = ImageDataGenerator(
        preprocessing_function=preprocess_func
    ).flow_from_directory(
        test_dir,
        target_size=size,
        batch_size=BATCH,
        class_mode="categorical",
        shuffle=False
    )

    # ---------------- MODEL ----------------
    base_model = base_model_class(
        include_top=False,
        weights="imagenet",
       input_shape=(*size,3)
    )

    base_model.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = BatchNormalization()(x)
    x = Dense(256, activation="relu")(x)
    x = Dropout(0.6)(x)
    output = Dense(5, activation="softmax")(x)

    model = Model(base_model.input, output)

    model.compile(
        optimizer="adam",
        loss=focal_loss(),
        metrics=["accuracy"]
    )

    callbacks = [
        EarlyStopping(patience=5, restore_best_weights=True, monitor="val_accuracy"),
        ReduceLROnPlateau(patience=2, factor=0.3)
    ]

    start = time.time()

    # ---------------- STAGE 1 ----------------
    history1 = model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=EPOCHS_STAGE1,
        callbacks=callbacks,
        class_weight=class_weights
    )

    # ---------------- STAGE 2 ----------------
    base_model.trainable = True

    for layer in base_model.layers[:-120]:
        layer.trainable = False

    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        #loss=focal_loss(),
        loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
        metrics=["accuracy"]
    )

    history2 = model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=EPOCHS_STAGE2,
        callbacks=callbacks
    )

    loss, acc = model.evaluate(test_gen)

    end = time.time()
    training_time = round((end - start)/60, 2)

    print(f"{model_name} Test Accuracy: {acc*100:.2f}%")
    print(f"Training Time: {training_time} minutes")

    return model, acc, training_time, history1, history2
```

```python
model_eff, acc_eff, time_eff, h1_eff, h2_eff = train_model("EfficientNetB3", EfficientNetB3, eff_pre)
model_res, acc_res, time_res, h1_res, h2_res = train_model("ResNet50", ResNet50, res_pre)
model_dense, acc_dense, time_dense, h1_dense, h2_dense = train_model("DenseNet121", DenseNet121, dense_pre)
```

```python
import matplotlib.pyplot as plt

def plot_history(history1, history2, model_name):

    acc = history1.history['accuracy'] + history2.history['accuracy']
    val_acc = history1.history['val_accuracy'] + history2.history['val_accuracy']

    loss = history1.history['loss'] + history2.history['loss']
    val_loss = history1.history['val_loss'] + history2.history['val_loss']

    epochs = range(len(acc))

    plt.figure(figsize=(12,5))

    # Accuracy
    plt.subplot(1,2,1)
    plt.plot(epochs, acc, label="Train Accuracy")
    plt.plot(epochs, val_acc, label="Validation Accuracy")
    plt.title(f"{model_name} Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()

    # Loss
    plt.subplot(1,2,2)
    plt.plot(epochs, loss, label="Train Loss")
    plt.plot(epochs, val_loss, label="Validation Loss")
    plt.title(f"{model_name} Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()

    plt.show()


# Plot graphs for each model
plot_history(h1_eff, h2_eff, "EfficientNetB3")
plot_history(h1_res, h2_res, "ResNet50")
plot_history(h1_dense, h2_dense, "DenseNet121")
```

```python
print("\n================ ENSEMBLE EVALUATION =================")

# EfficientNet test generator
test_gen_eff = ImageDataGenerator(
    preprocessing_function=eff_pre
).flow_from_directory(
    test_dir,
    target_size=(300,300),
    batch_size=BATCH,
    class_mode="categorical",
    shuffle=False
)

# ResNet test generator
test_gen_res = ImageDataGenerator(
    preprocessing_function=res_pre
).flow_from_directory(
    test_dir,
    target_size=(224,224),
    batch_size=BATCH,
    class_mode="categorical",
    shuffle=False
)

# DenseNet test generator
test_gen_dense = ImageDataGenerator(
    preprocessing_function=dense_pre
).flow_from_directory(
    test_dir,
    target_size=(300,300),
    batch_size=BATCH,
    class_mode="categorical",
    shuffle=False
)

# Predictions
pred_eff = model_eff.predict(test_gen_eff, verbose=1)
pred_res = model_res.predict(test_gen_res, verbose=1)
pred_dense = model_dense.predict(test_gen_dense, verbose=1)

# Ensemble average
ensemble_pred = (pred_eff + pred_res + pred_dense) / 3

# True labels
y_true = test_gen_eff.classes

# Final prediction
y_pred = np.argmax(ensemble_pred, axis=1)

# Accuracy
from sklearn.metrics import accuracy_score
ensemble_acc = accuracy_score(y_true, y_pred)

print(f"\nEnsemble Accuracy: {ensemble_acc*100:.2f}%")
```

```python
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator

def tta_predict(model, test_dir, preprocess_func, img_size, batch_size, TTA=7):

    preds = []

    for i in range(TTA):

        tta_gen = ImageDataGenerator(
            preprocessing_function=preprocess_func,
            rotation_range=10,
            zoom_range=0.1,
            width_shift_range=0.05,
            height_shift_range=0.05
        ).flow_from_directory(
            test_dir,
            target_size=img_size,
            batch_size=batch_size,
            class_mode="categorical",
            shuffle=False
        )

        pred = model.predict(tta_gen, verbose=0)

        preds.append(pred)

    preds = np.mean(preds, axis=0)

    return preds
```

```python
from sklearn.metrics import accuracy_score

def evaluate_tta(model, preprocess, img_size, name):

    pred_probs = tta_predict(
        model,
        test_dir,
        preprocess,
        img_size,
        BATCH,
        TTA=7
    )

    y_pred = np.argmax(pred_probs, axis=1)

    test_gen = ImageDataGenerator(
        preprocessing_function=preprocess
    ).flow_from_directory(
        test_dir,
        target_size=img_size,
        batch_size=BATCH,
        class_mode="categorical",
        shuffle=False
    )

    y_true = test_gen.classes

    acc = accuracy_score(y_true, y_pred)

    print(f"{name} TTA Accuracy: {acc*100:.2f}%")
```

```python
print("\nTTA Evaluation Results\n")

evaluate_tta(model_eff, eff_pre, IMG_SIZE, "EfficientNetB3")
evaluate_tta(model_res, res_pre, (224,224), "ResNet50")
evaluate_tta(model_dense, dense_pre, IMG_SIZE, "DenseNet121")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

def get_predictions(model, generator):
    generator.reset()
    preds = model.predict(generator, verbose=1)
    y_pred = np.argmax(preds, axis=1)
    y_true = generator.classes
    return y_true, y_pred
```

```python
def plot_conf_matrix(y_true, y_pred, model_name, class_names):

    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(6,5))
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
    disp.plot(cmap="Blues", values_format='d')

    plt.title(f"{model_name} Confusion Matrix")
    plt.xticks(rotation=45)
    plt.show()
```
